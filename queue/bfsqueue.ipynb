# Biggest Queue Application: Breadth First Search

## Idea: Moving Level by Level:

## Using deque Class for Queue:

```python
from collections import deque
```

## Graph Structure: with adjacency list:

```python
graph = {
    'A':['B','C'],
    'B':['D','E'],
    'C':[],
    'D':[],
    'E':[]
}
```

```python
print(graph)
```

## Breadth First Search: BFS:

```python
def bfs(gr,start):
    queue = deque([start])

    visited = set()

    while queue:
        node = queue.popleft()
        print(node)

        for neighbour in gr[node]:
            if neighbour not in visited:
                visited.add(neighbour)
                queue.append(neighbour)
```

```python
result = bfs(graph, 'A')
result 
# Output: ['A', 'B', 'C', 'D', 'E']
```

# Task:
- Modify BFS to return traversal list, not print.

```python
def bfswithreturn(gr,start):

    queue = deque([start])
    visited = {start}
    l = []
    
    while queue:
        node = queue.popleft()
        l.append(node)
        for neighbour in gr[node]:
            if neighbour not in visited:
                visited.add(neighbour)
                queue.append(neighbour)
    return l
```

```python
result = bfswithreturn(graph, 'A')
print(result)
```
