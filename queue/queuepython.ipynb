# Python Way of Implementing Queue: deque

```python
from collections import deque
```

```python
deque
```

Efficient at insertion and removal:

- front
- back

# Basic Queue:

```python
queue = deque()
```

```python
queue.append(1)
queue.append(2)
queue.append(3)
queue.append(4)
queue.append(5)

print(f"Queue: {queue}")
```

```python
print(f"Removed: {queue.popleft()}")

print(f"Queue: {queue}")
```

# Methods:

- append(x)

- appendleft(x)

- pop()

- popleft()

```python
queue.appendleft(7)

print(f"Queue: {queue}")
```

```python
print(f"Removed (from right): {queue.pop()}")

print(f"Queue: {queue}")
```
