# Task: Queue using 2 Stacks:

```python
class Stack:
    def __init__(self) -> None:
        self.stack = []

    def push(self,value):
        self.stack.append(value)
        
    def is_empty(self):
        return (len(self.stack) == 0)    

    def pop(self):
        if self.is_empty():
            raise IndexError("Stack is Empty!")
        return self.stack.pop()
    
    def peek(self):
        if self.is_empty():
            return None
        return self.stack[-1]
    
    def size(self):
        return len(self.stack)

    def print_stack(self):
        if self.is_empty():
            print("Stack is Empty!")
            return
        for i in range(self.size() - 1, -1, -1):
            display_item = str(self.stack[i])
            print(f"|{display_item}|")
```

```python
class Queue2Stack:

    def __init__(self) -> None:
        self.instack = Stack()
        self.outstack = Stack()

    def enqueue(self,value):
        self.instack.push(value)

    def dequeue(self):
        if self.instack.is_empty() and self.outstack.is_empty():
            raise IndexError("Queue is empty!")

        if self.outstack.is_empty():
            while not self.instack.is_empty():
                self.outstack.push(self.instack.pop())

        return self.outstack.pop()

    def front(self):
        if self.is_empty():
            return None
        return self.outstack.peek()
        
    def is_empty(self):
        return self.instack.is_empty() and self.outstack.is_empty()

    def size(self):
        return self.instack.size() + self.outstack.size()
    
    def print_queue(self):
        if self.is_empty():
            print("Queue is empty")
            return
        
        # 1. Elements in out_stack are reversed (top of stack is front of queue).
        # So we read out_stack backwards to get front-to-back order.
        front_part = self.outstack.stack[::-1]
        
        # 2. Elements in in_stack are in normal order (bottom of stack is older).
        # So we read in_stack normally from bottom to top.
        back_part = self.instack.stack
        
        # 3. Combine them to get the full queue in correct FIFO order!
        full_queue = front_part + back_part
        
        # Print it nicely
        print("Front -> " + " -> ".join(map(str, full_queue)) + " <- Back")
```

```python
q = Queue2Stack()

print(f"\nFront of Queue: {q.front()}")
print(f"Size of Queue: {q.size()}")

q.enqueue(1)
q.enqueue(2)
q.enqueue(3)
q.enqueue(4)
q.enqueue(5)

print("\nQueue: ",end="")
q.print_queue()

print(f"dequeue: {q.dequeue()}")
print(f"dequeue: {q.dequeue()}")


print("\nQueue: ",end="")
q.print_queue()

print(f"\nFront of Queue: {q.front()}")
print(f"Size of Queue: {q.size()}")
```
