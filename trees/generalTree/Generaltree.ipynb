# General Tree:

A tree with no constraints on how many children a node can have (e.g., a file directory system).

```python
from generalnode import GeneralNode
from typing import Optional
```

```python
class GeneralTree:
    def __init__(self,root:GeneralNode) -> None:
        self.root = root

    def add_child(self,parent_node:GeneralNode,child_node:GeneralNode):
        if parent_node is None:
            return
        parent_node.children.append(child_node)
    
    # preorder:
    def traverse_dfs(self,node:GeneralNode):
        if node is None:
            return
        print(node.data)
        for i in node.children:
            self.traverse_dfs(i)

    def print_relationships(self, node: GeneralNode, parent: Optional[GeneralNode] = None):
        if node is None:
            return
        parent_data = parent.data if parent else "None"
        
        children_data = [child.data for child in node.children]
        if not children_data:
            children_data = ["None"]
            
        print(f"Node({node.data}) -> Parent: {parent_data} | Children: {children_data}")
        
        for child in node.children:
            self.print_relationships(child, parent=node)
```

```python
# --- Running the Fixed Test Code ---

a = GeneralNode(1)
general = GeneralTree(a)  # Passes the actual root node

b = GeneralNode(2)
c = GeneralNode(3)
d = GeneralNode(4)
e = GeneralNode(5)

general.add_child(a, b)   # Adds 2 as child of 1
general.add_child(a, c)   # Adds 3 as child of 1
general.add_child(b, d)   # Adds 4 as child of 2
general.add_child(b, e)   # Adds 5 as child of 2
```
