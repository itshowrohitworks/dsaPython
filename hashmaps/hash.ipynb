# Initialization and data:

## 1. Initialize:

```python
fruits = {
    "apple":10
}
print(fruits)
```

## 2. Accessing:

```python
fruits["apple"]
```

## 3. Insert:

```python
fruits["banana"] = 5
print(fruits)
```

## 4. Delete:

```python
del fruits["banana"]
print(fruits)
```

# Common Methods:

```python
student = {
    "name":"Something",
    "age":55
}
```

## 1. Keys: get all keys in the dict

```python
student.keys()
```

## 2. Values: get all values in the dict

```python
student.values()
```

## 3. items(): returns key-value pairs in tuple format

```python
student.items()
```

## 4. Looping: Python unpacking happens automatically

```python
for key,value in student.items():
    print(key,value)
```

# Sets:
- sets are hash tables too.
- Fast lookup.

Why set matters?
- Set lookup is O(1)

Real World use:
- duplicate removal
- recommendation systems
- visited nodes in graphs
- caching
- fraud detection
- membership testing

```python
emails = set()

emails.add("a@gmail.com")

if "a@gmail.com" in emails:
    print("Exists")
else:
    print("Not Exists")
```

# BUILT-IN TOOLS:

## 1. Counter:
- Used for:
NLP,
ML preprocessing,
analytics,
frequency counting.

```python
from collections import Counter


arr = [1,1,2,3,3,3]

count = Counter(arr)

print(count)
```

# defaultdict
- Avoids checking existence manually like the normal dict in python.
- Cleaner and faster to write.

```python
from collections import defaultdict

d = defaultdict(list)

d["a"].append(10)

print(d)
```
